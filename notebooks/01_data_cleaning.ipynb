```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("hotels.csv")
```

```python
print("Shape:", df.shape)
print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

```python
df_clean = df.copy()
```

```python
df_clean.info()
```

```python
df_clean.describe()
```

```python
df_clean.describe(include="object")
```

```python
missing = pd.DataFrame({
    "Missing_Count": df_clean.isnull().sum(),
    "Missing_Percentage": (
        df_clean.isnull().mean() * 100
    )
})

missing = missing.sort_values(
    "Missing_Percentage",
    ascending=False
)

missing
```

```python
for col in ["agent", "company"]:

    df_clean[col] = (
        pd.to_numeric(df_clean[col], errors="coerce")
        .astype("Int64")
        .astype("string")
        .fillna("None")
    )
```

```python
df_clean["country"] = (
    df_clean["country"].fillna("Unknown")
)
```

```python
print(df_clean["children"].isnull().sum())

print(df_clean.loc[
    df_clean["children"].isnull()
])
```

```python
df_clean["children"] = (
    df_clean["children"].fillna(0)
)
```

```python
print("Duplicate rows:",
      df_clean.duplicated().sum())
```

```python
assert len(df_clean) == 119390
```

```python
df_clean["reservation_status_date"] = pd.to_datetime(
    df_clean["reservation_status_date"],
    errors="coerce"
)
```

```python
print(df_clean["reservation_status_date"].dtype)
```

```python
bin3_leakage_cols = [
    "reservation_status",
    "reservation_status_date",
    "reservation_year",
    "assigned_room_type"
]
```

```python
bin2_amendable_cols = [
    "booking_changes",
    "days_in_waiting_list",
    "required_car_parking_spaces",
    "total_of_special_requests"
]
```

```python
columns_to_drop = (
    bin3_leakage_cols +
    bin2_amendable_cols
)

model_df = df_clean.drop(
    columns=columns_to_drop,
    errors="ignore"
).copy()
```

```python
print("Original cleaned data:", df_clean.shape)
print("Modelling data:", model_df.shape)
```

```python
remaining = [
    col for col in columns_to_drop
    if col in model_df.columns
]

print("Prohibited columns remaining:", remaining)
```

```python
model_df["adr"] = model_df["adr"].clip(lower=0)
```

```python
print("Minimum ADR:", model_df["adr"].min())
```

```python
model_df["nights"] = (
    model_df["stays_in_weekend_nights"] +
    model_df["stays_in_week_nights"]
)
```

```python
model_df["total_guests"] = (
    model_df["adults"] +
    model_df["children"] +
    model_df["babies"]
)
```

```python
model_df["arrival_date"] = pd.to_datetime(
    model_df["arrival_date_year"].astype(str) + "-" +
    model_df["arrival_date_month"] + "-" +
    model_df["arrival_date_day_of_month"].astype(str),
    format="%Y-%B-%d",
    errors="coerce"
)
```

```python
sns.boxplot(x=model_df["adr"])

plt.title("ADR Outlier Analysis")
plt.show()
```

```python
Q1 = model_df["adr"].quantile(0.25)
Q3 = model_df["adr"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = model_df[
    (model_df["adr"] < lower) |
    (model_df["adr"] > upper)
]

print("Outliers:", len(outliers))
```

```python
print("Number of rows:", len(model_df))

print("Number of columns:", len(model_df.columns))

print("Missing values:", model_df.isnull().sum().sum())

print("Duplicate rows:", model_df.duplicated().sum())

print("Minimum ADR:", model_df["adr"].min())

print("Data types:")
print(model_df.dtypes)
```

```python
# All booking records preserved
assert len(df_clean) == 119390

# No negative ADR
assert model_df["adr"].ge(0).all()

# Target is valid
assert model_df["is_canceled"].isin([0, 1]).all()

# Leakage columns removed
assert not set(bin3_leakage_cols).intersection(model_df.columns)

# Amendable columns removed
assert not set(bin2_amendable_cols).intersection(model_df.columns)

# IDs stored as strings
assert model_df["agent"].dtype == "string"
assert model_df["company"].dtype == "string"

print("All project-rule checks passed!")
```

```python
df_clean.to_csv(
    "hotels_cleaned.csv",
    index=False
)
```

```python
model_df.to_csv(
    "hotels_model_ready.csv",
    index=False
)
```
